**Prototype EXCEL**

### Upload your .txt file

```python
from google.colab import files
import pandas as pd

# Upload the .txt file
uploaded = files.upload()

# Get the filename (assuming only one file is uploaded)
for fn in uploaded.keys():
  print(f'User uploaded file "{fn}"')
  text_filename = fn
```

### Convert .txt to Excel

```python
# .txt to Excel
try:
    df = pd.read_csv(text_filename, sep='\t', header=0)
    print("Text file successfully read into DataFrame.")
except Exception as e:
    print(f"Error reading text file: {e}")
    print("Please check the file format (e.g., delimiter) and try again.")
    df = pd.DataFrame() # Create an empty DataFrame to avoid errors later

# Define the output Excel filename for the full dataset
excel_filename = text_filename.replace('.txt', '.xlsx') if '.txt' in text_filename else text_filename + '.xlsx'

# Write the DataFrame to an Excel file
if not df.empty:
    df.to_excel(excel_filename, index=False)
    print(f"Full DataFrame successfully written to {excel_filename}")
    display(df.head())
else:
    print("DataFrame is empty, cannot create full Excel file.")
```

### Select and save specific columns

```python
# Identify the columns to select by their names
selected_columns_names = ['Age', 'Gender', 'Credit_Score', 'Loan_Term_Months', 'Interest_Rate', 'Marital_Status']

# Check if all selected columns exist in the DataFrame
missing_columns = [col for col in selected_columns_names if col not in df.columns]

if not missing_columns:
    # Select only the specified columns
    df_selected = df[selected_columns_names]
    print("Selected columns successfully:")
    display(df_selected.head())

    # Define the output Excel filename for the selected data
    selected_excel_filename = text_filename.replace('.txt', '_selected.xlsx') if '.txt' in text_filename else text_filename + '_selected.xlsx'

    # Write the selected DataFrame to a new Excel file
    df_selected.to_excel(selected_excel_filename, index=False)
    print(f"Selected columns successfully written to {selected_excel_filename}")

    # Provide a download link for the new Excel file
    from google.colab import files
    files.download(selected_excel_filename)
    print(f"Your Excel file with selected columns '{selected_excel_filename}' is ready for download.")
else:
    print(f"Error: The following columns were not found in the DataFrame: {missing_columns}")
    print("Please check the column names in your .txt file.")
```

### Download the Excel file

```python
# Provide a download link for the Excel file
if 'excel_filename' in locals() and pd.notna(excel_filename) and not df.empty:
    files.download(excel_filename)
    print(f"Your converted Excel file '{excel_filename}' is ready for download.")
else:
    print("No Excel file was generated or found to download.")
```
